# 📊 Metodología del Análisis

## 1. Comprensión del dataset
Se realizó una exploración inicial para comprender la estructura de los datos, identificar las variables disponibles y estudiar el significado de cada columna.

---

## 2. Limpieza y preparación de datos
Durante esta fase se:

- Eliminaron columnas redundantes.
- Creó una columna de fecha a partir del año y el mes.
- Revisaron valores nulos y posibles inconsistencias.
- Clasificaron los productos energéticos según su naturaleza.

---

## 3. Ingeniería de características
Se añadieron nuevas variables para facilitar el análisis:

### CATEGORY
Clasifica cada producto energético como:

- Source
- Aggregate
- Balance

### RENEWABLE
Clasifica las fuentes energéticas en:

- Renewable
- Non-renewable
- Mixed / Other

### CONTINENT
Agrupa los países por continente para facilitar comparaciones regionales.

### GDP
Incorpora el PIB de cada país para analizar posibles relaciones entre desarrollo económico y producción energética.

---

## 4. Análisis exploratorio
Se realizaron distintos análisis para responder a la hipótesis planteada:

- Evolución temporal de las energías renovables.
- Crecimiento renovable por país.
- Producción energética total por país.
- Distribución porcentual entre energías renovables y no renovables.
- Relación entre PIB y producción energética.
- Producción renovable relativa al PIB.
- Comparaciones entre continentes.

---

## 5. Visualización de datos
Se construyeron gráficos para identificar patrones, tendencias y relaciones entre variables relevantes.

Entre ellos:

- Producción energética total por país.
- Evolución de fuentes renovables.
- Producción renovable relativa al PIB.
- Distribución energética por categorías.
- Comparativas continentales.

---

## 6. Obtención de conclusiones
Finalmente se analizaron los resultados para evaluar si la transición hacia energías renovables depende únicamente de factores económicos o si otros factores, como las políticas energéticas, tienen un papel determinante.
